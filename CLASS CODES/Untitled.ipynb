```python
import pandas as pd
import json
from bs4 import BeautifulSoup
import requests

quotes = []
authors = []

url = "https://quotes.toscrape.com"
response = requests.get(url)

# 1. Check if request succeeded
if response.status_code != 200:
    print("Failed to fetch page")
    exit()

# 2. Parse the HTML into a BeautifulSoup object
soup = BeautifulSoup(response.text, "html.parser")

# 3. Now extract quotes and authors
for quote in soup.find_all("div", class_="quote"):
    text = quote.find("span", class_="text").get_text()
    author = quote.find("small", class_="author").get_text()
    quotes.append(text)
    authors.append(author)

# 4. Print results
print("All Quotes:", quotes)
print("All Authors:", authors)

# (Optional) Create DataFrame
df = pd.DataFrame({"Quote": quotes, "Author": authors})
print(df)



csv = df.to_csv('quotes.csv', index=False)

json = df.to_json('quotes.json', orient='records', indent=2)
```
