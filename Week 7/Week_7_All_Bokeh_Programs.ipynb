# Program 13 – Interactive Scatter Plot with Bokeh

**Problem Statement:**  
Create an interactive scatter plot showing the relationship between petal length and petal width from the Iris dataset using Bokeh.

**Main tasks:**
- Load the built-in Iris dataset.
- Use petal length, petal width and species.
- Calculate species counts and correlation.
- Create a Bokeh scatter plot.
- Add hover and interactive legend.
- Save the chart as an HTML file.


```python
import pandas as pd
from sklearn.datasets import load_iris
from bokeh.plotting import figure, output_file, save, show
from bokeh.models import ColumnDataSource, HoverTool
from bokeh.palettes import Category10
from bokeh.io import output_notebook

output_notebook()

# Load the Iris dataset
iris = load_iris(as_frame=True)

df = iris.frame.rename(columns={
    "petal length (cm)": "Petal Length (cm)",
    "petal width (cm)": "Petal Width (cm)"
})

df["Species"] = df["target"].map(dict(enumerate(iris.target_names)))
df = df[["Petal Length (cm)", "Petal Width (cm)", "Species"]]

print(df.head(10))
print("\nDataset Shape:", df.shape)
print("\nSpecies Counts:")
print(df["Species"].value_counts())

correlation = df["Petal Length (cm)"].corr(df["Petal Width (cm)"])
print("\nCorrelation:", round(correlation, 3))
```

```python
# Create the interactive Bokeh scatter plot
p = figure(
    title="Interactive Scatter Plot: Iris Petal Length vs Petal Width",
    x_axis_label="Petal Length (cm)",
    y_axis_label="Petal Width (cm)",
    width=900,
    height=520,
    tools="pan,wheel_zoom,box_zoom,reset,save"
)

colors = Category10[3]

for color, species in zip(colors, iris.target_names):
    subset = df[df["Species"] == species]
    source = ColumnDataSource(subset)

    p.scatter(
        x="Petal Length (cm)",
        y="Petal Width (cm)",
        source=source,
        size=9,
        color=color,
        alpha=0.75,
        legend_label=species
    )

hover = HoverTool(tooltips=[
    ("Species", "@Species"),
    ("Petal Length", "@{Petal Length (cm)}{0.0} cm"),
    ("Petal Width", "@{Petal Width (cm)}{0.0} cm")
])

p.add_tools(hover)
p.legend.title = "Species"
p.legend.location = "top_left"
p.legend.click_policy = "hide"

output_file("program13_iris_scatter.html")
save(p)
show(p)

print("Interactive scatter plot saved successfully.")
```

# Program 14 – Interactive Line Plot with Bokeh

**Problem Statement:**  
Create an interactive line plot showing the daily temperatures over a week using Bokeh.

**Main tasks:**
- Create a seven-day temperature dataset.
- Calculate highest, lowest and average temperature.
- Plot the daily trend using a Bokeh line plot.
- Add markers and hover information.
- Save the chart as an HTML file.


```python
import pandas as pd
from bokeh.plotting import figure, output_file, save, show
from bokeh.models import ColumnDataSource, HoverTool
from bokeh.io import output_notebook

output_notebook()

# Weekly temperature dataset
days = [
    "Monday", "Tuesday", "Wednesday",
    "Thursday", "Friday", "Saturday", "Sunday"
]

temperatures = [29, 31, 30, 32, 34, 33, 31]

df_temp = pd.DataFrame({
    "Day": days,
    "Temperature (°C)": temperatures
})

print(df_temp)

highest_row = df_temp.loc[df_temp["Temperature (°C)"].idxmax()]
lowest_row = df_temp.loc[df_temp["Temperature (°C)"].idxmin()]
average_temp = df_temp["Temperature (°C)"].mean()

print("\nHighest Temperature:",
      highest_row["Day"], "-",
      highest_row["Temperature (°C)"], "°C")

print("Lowest Temperature:",
      lowest_row["Day"], "-",
      lowest_row["Temperature (°C)"], "°C")

print("Average Temperature:",
      round(average_temp, 2), "°C")
```

```python
source = ColumnDataSource(df_temp)

p = figure(
    x_range=days,
    title="Interactive Line Plot: Daily Temperature Over a Week",
    x_axis_label="Day",
    y_axis_label="Temperature (°C)",
    width=900,
    height=520,
    tools="pan,wheel_zoom,box_zoom,reset,save"
)

p.line(
    x="Day",
    y="Temperature (°C)",
    source=source,
    line_width=3
)

p.scatter(
    x="Day",
    y="Temperature (°C)",
    source=source,
    size=10
)

hover = HoverTool(tooltips=[
    ("Day", "@Day"),
    ("Temperature", "@{Temperature (°C)} °C")
])

p.add_tools(hover)
p.xgrid.grid_line_color = None

output_file("program14_temperature_line.html")
save(p)
show(p)

print("Interactive line plot saved successfully.")
```

# Program 15 – Interactive Bar Chart with Bokeh

**Problem Statement:**  
Create an interactive bar chart showing the sales figures of different products using Bokeh.

**Main tasks:**
- Create a product-wise sales dataset.
- Calculate total, highest and average sales.
- Create an interactive Bokeh vertical bar chart.
- Add hover information.
- Save the chart as an HTML file.


```python
import pandas as pd
from bokeh.plotting import figure, output_file, save, show
from bokeh.models import ColumnDataSource, HoverTool
from bokeh.io import output_notebook

output_notebook()

# Product sales dataset
products = [
    "Laptop", "Smartphone", "Tablet",
    "Headphones", "Smartwatch", "Camera"
]

sales = [125, 210, 95, 170, 140, 80]

df_sales = pd.DataFrame({
    "Product": products,
    "Sales (Units)": sales
})

print(df_sales)

highest_row = df_sales.loc[df_sales["Sales (Units)"].idxmax()]
total_sales = df_sales["Sales (Units)"].sum()
average_sales = df_sales["Sales (Units)"].mean()

print("\nTotal Sales:", total_sales, "units")
print("Highest Sales:",
      highest_row["Product"], "-",
      highest_row["Sales (Units)"], "units")
print("Average Sales:", round(average_sales, 2), "units")
```

```python
source = ColumnDataSource(df_sales)

p = figure(
    x_range=products,
    title="Interactive Bar Chart: Product Sales",
    x_axis_label="Product",
    y_axis_label="Sales (Units)",
    width=900,
    height=520,
    tools="pan,wheel_zoom,box_zoom,reset,save"
)

p.vbar(
    x="Product",
    top="Sales (Units)",
    source=source,
    width=0.65
)

hover = HoverTool(tooltips=[
    ("Product", "@Product"),
    ("Sales", "@{Sales (Units)} units")
])

p.add_tools(hover)
p.xgrid.grid_line_color = None

output_file("program15_product_sales_bar.html")
save(p)
show(p)

print("Interactive bar chart saved successfully.")
```
