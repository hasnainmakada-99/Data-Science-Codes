```python
import pandas as pd
from bokeh.plotting import figure, output_file, save
from bokeh.models import ColumnDataSource, HoverTool
# Weekly temperature dataset
days = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday",
"Saturday", "Sunday"]
temperatures = [29, 31, 30, 32, 34, 33, 31]
df = pd.DataFrame({"Day": days, "Temperature (°C)": temperatures})
print(df)
highest_row = df.loc[df["Temperature (°C)"].idxmax()]
lowest_row = df.loc[df["Temperature (°C)"].idxmin()]
average_temp = df["Temperature (°C)"].mean()
print("Highest Temperature:", highest_row["Day"], "-",
highest_row["Temperature (°C)"], "°C")
print("Lowest Temperature:", lowest_row["Day"], "-",
lowest_row["Temperature (°C)"], "°C")
print("Average Temperature:", round(average_temp, 2), "°C")
source = ColumnDataSource(df)
p = figure(
x_range=days,
title="Interactive Line Plot: Daily Temperature Over a Week",
x_axis_label="Day",
y_axis_label="Temperature (°C)",
width=900,
height=520,
tools="pan,wheel_zoom,box_zoom,reset,save"
)

p.line(x="Day", y="Temperature (°C)", source=source, line_width=3)
p.scatter(x="Day", y="Temperature (°C)", source=source, size=10)
hover = HoverTool(tooltips=[
("Day", "@Day"),
("Temperature", "@{Temperature (°C)} °C")
])
p.add_tools(hover)
p.xgrid.grid_line_color = None

output_file("program14_temperature_line.html")
save(p)
print("Interactive line plot saved successfully.")
```
