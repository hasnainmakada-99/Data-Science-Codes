# Interactive Bar Chart with Plotly

## Week 6 Practice Program

### Problem Statement
Create an interactive bar chart showing the population of different cities.

### Dataset
The dataset contains the population of six major Indian cities in lakhs.

```python
import pandas as pd
import plotly.express as px
```

```python
data = {
    "City": [
        "Mumbai",
        "Delhi",
        "Bengaluru",
        "Hyderabad",
        "Chennai",
        "Kolkata"
    ],
    
    "Population (Lakhs)": [
        204,
        167,
        136,
        108,
        115,
        149
    ]
}
```

```python
df = pd.DataFrame(data)

df
```

```python
print("Dataset Shape:", df.shape)

print("\nCities:")
print(df["City"].tolist())

print("\nPopulation:")
print(df["Population (Lakhs)"].tolist())
```

```python
fig = px.bar(
    df,
    x="City",
    y="Population (Lakhs)",
    title="Population of Major Indian Cities",
    text="Population (Lakhs)"
)

fig.update_traces(
    textposition="outside"
)

fig.update_layout(
    xaxis_title="City",
    yaxis_title="Population (Lakhs)"
)

fig.show()
```

```python
sorted_df = df.sort_values(
    by="Population (Lakhs)",
    ascending=False
)

sorted_df
```

```python
fig = px.bar(
    sorted_df,
    x="City",
    y="Population (Lakhs)",
    title="City-wise Population Analysis",
    text="Population (Lakhs)"
)

fig.update_traces(
    textposition="outside"
)

fig.update_layout(
    xaxis_title="City",
    yaxis_title="Population (Lakhs)"
)

fig.show()
```

```python
highest = df.loc[
    df["Population (Lakhs)"].idxmax()
]

print("City with Highest Population:")
print(highest)
```

```python
lowest = df.loc[
    df["Population (Lakhs)"].idxmin()
]

print(
    "Lowest Population:",
    lowest["City"],
    "-",
    lowest["Population (Lakhs)"],
    "Lakhs"
)
```

```python
average_population = df["Population (Lakhs)"].mean()

print(
    "Average Population:",
    round(average_population, 2),
    "Lakhs"
)
```

```python
print("CITY POPULATION ANALYSIS")
print("-" * 35)

print("Total Cities:", len(df))

print(
    "Highest Population:",
    df.loc[df["Population (Lakhs)"].idxmax(), "City"]
)

print(
    "Lowest Population:",
    df.loc[df["Population (Lakhs)"].idxmin(), "City"]
)

print(
    "Average Population:",
    round(df["Population (Lakhs)"].mean(), 2),
    "Lakhs"
)
```

```python
fig.write_html(
    "city_population_chart.html"
)

print("Interactive chart saved successfully.")
```

```python
df.to_csv(
    "city_population.csv",
    index=False
)

print("Dataset saved successfully.")
```
