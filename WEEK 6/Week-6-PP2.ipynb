# Interactive Scatter Plot with Plotly

## Week 6 Practice Program

### Problem Statement
Create an interactive scatter plot showing the relationship between house size and price.

### Dataset
The dataset contains house size in square feet and corresponding house price in lakhs.

```python
import pandas as pd
import plotly.express as px
```

```python
data = {
    "House Size (sq.ft.)": [
        800,
        1000,
        1200,
        1400,
        1600,
        1800,
        2000,
        2200,
        2500,
        2800
    ],

    "House Price (Lakhs)": [
        35,
        45,
        55,
        65,
        78,
        90,
        105,
        120,
        140,
        165
    ]
}
```

```python
df = pd.DataFrame(data)

df
```

```python
print("Dataset Shape:", df.shape)

print("\nHouse Sizes:")
print(df["House Size (sq.ft.)"].tolist())

print("\nHouse Prices:")
print(df["House Price (Lakhs)"].tolist())
```

```python
fig = px.scatter(
    df,
    x="House Size (sq.ft.)",
    y="House Price (Lakhs)",
    title="House Size vs House Price"
)

fig.show()
```

```python
fig = px.scatter(
    df,
    x="House Size (sq.ft.)",
    y="House Price (Lakhs)",
    title="Relationship Between House Size and Price",
    text="House Price (Lakhs)"
)

fig.update_traces(
    marker=dict(size=12),
    textposition="top center"
)

fig.update_layout(
    xaxis_title="House Size (sq.ft.)",
    yaxis_title="House Price (₹ Lakhs)"
)

fig.show()
```

```python
sorted_df = df.sort_values(
    by="House Size (sq.ft.)",
    ascending=True
)

sorted_df
```

```python
highest = df.loc[
    df["House Price (Lakhs)"].idxmax()
]

print("Most Expensive House:")
print(highest)
```

```python
lowest = df.loc[
    df["House Price (Lakhs)"].idxmin()
]

print(
    "Least Expensive House:",
    lowest["House Size (sq.ft.)"],
    "sq.ft. - ₹",
    lowest["House Price (Lakhs)"],
    "Lakhs"
)
```

```python
average_price = df["House Price (Lakhs)"].mean()

print(
    "Average House Price:",
    round(average_price, 2),
    "Lakhs"
)
```

```python
average_size = df["House Size (sq.ft.)"].mean()

print(
    "Average House Size:",
    round(average_size, 2),
    "sq.ft."
)
```

```python
correlation = df[
    "House Size (sq.ft.)"
].corr(
    df["House Price (Lakhs)"]
)

print(
    "Correlation between House Size and Price:",
    round(correlation, 3)
)
```

```python
fig = px.scatter(
    df,
    x="House Size (sq.ft.)",
    y="House Price (Lakhs)",
    title="House Size vs Price with Trendline",
    trendline="ols"
)

fig.update_traces(
    marker=dict(size=12)
)

fig.update_layout(
    xaxis_title="House Size (sq.ft.)",
    yaxis_title="House Price (₹ Lakhs)"
)

fig.show()
```

```python
print("HOUSE PRICE ANALYSIS")
print("-" * 40)

print("Total Houses:", len(df))

print(
    "Smallest House:",
    df["House Size (sq.ft.)"].min(),
    "sq.ft."
)

print(
    "Largest House:",
    df["House Size (sq.ft.)"].max(),
    "sq.ft."
)

print(
    "Lowest Price: ₹",
    df["House Price (Lakhs)"].min(),
    "Lakhs"
)

print(
    "Highest Price: ₹",
    df["House Price (Lakhs)"].max(),
    "Lakhs"
)

print(
    "Average Price: ₹",
    round(df["House Price (Lakhs)"].mean(), 2),
    "Lakhs"
)

print(
    "Correlation:",
    round(correlation, 3)
)
```

```python
fig.write_html(
    "house_size_price_scatter.html"
)

print("Interactive scatter plot saved successfully.")
```

```python
df.to_csv(
    "house_price_data.csv",
    index=False
)

print("Dataset saved successfully.")
```
