```python
import pandas as pd
import numpy as np
from sklearn.cluster import KMeans
import plotly.express as px
```

# Dataset

```python
data = {
'Subject': [1, 2, 3, 4, 5, 6, 7],
'A': [1.0, 1.5, 3.0, 5.0, 3.5, 4.5, 3.5],
'B': [1.0, 2.0, 4.0, 7.0, 5.0, 5.0, 4.5]
}
df = pd.DataFrame(data)
```

# Features

```python
X = df[['A', 'B']]
```

# Initial Centroids

```python
initial_centroids = np.array([[1.0, 1.0], [5.0, 7.0]])
```

# K-Means

```python
kmeans = KMeans(n_clusters=2, init=initial_centroids, n_init=1, random_state=42)
kmeans.fit(X)
```

# Cluster Labels

```python
df['Cluster'] = kmeans.labels_
```

# Final Centroids

```python
print('Final Centroids')
print(kmeans.cluster_centers_)
```

# Create Z-Axis

```python
df['Z'] = df['Subject']
```

# Interactive 3D Plot

```python
fig = px.scatter_3d(df, x='A', y='B', z='Z', color=df['Cluster'].astype(str), text='Subject', hover_data=['Subject', 'A', 'B', 'Cluster'], title='K-Means Clustering Visualization')
fig.update_layout(
scene=dict( xaxis_title='Feature A', yaxis_title='Feature B', zaxis_title='Subject'))

fig.show()
```
