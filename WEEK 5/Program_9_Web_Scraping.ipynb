# Program 9: Web Scraping and Data Collection

Scrape quotes, authors, and tags from `quotes.toscrape.com`, organize the data with Pandas, and save it as CSV and JSON.

```python
import json
import pandas as pd
import requests
from bs4 import BeautifulSoup

url = 'https://quotes.toscrape.com/'
response = requests.get(url, timeout=30)
response.raise_for_status()
soup = BeautifulSoup(response.text, 'html.parser')
print('Website title:', soup.title.get_text(strip=True))
print('Status code:', response.status_code)
```

```python
quotes = []
for item in soup.find_all('div', class_='quote'):
    quotes.append({
        'Quote': item.find('span', class_='text').get_text(strip=True),
        'Author': item.find('small', class_='author').get_text(strip=True),
        'Tags': ', '.join(tag.get_text(strip=True) for tag in item.find_all('a', class_='tag'))
    })

df = pd.DataFrame(quotes)
display(df)
```

```python
df.to_csv('quotes.csv', index=False)
with open('quotes.json', 'w', encoding='utf-8') as file:
    json.dump(quotes, file, indent=4, ensure_ascii=False)

print('Data saved successfully in CSV and JSON format.')
display(pd.read_csv('quotes.csv'))
with open('quotes.json', encoding='utf-8') as file:
    display(pd.DataFrame(json.load(file)))
```
