```python
import pandas as pd
import plotly.express as px

data = {'City': ['Mumbai', 'Delhi', 'Bengaluru', 'Hyderabad', 'Chennai', 'Kolkata'],
        'Population (Lakhs)': [204, 167, 136, 108, 115, 149]}
df = pd.DataFrame(data)

fig = px.bar(df, x='City', y='Population (Lakhs)', color='City',
             title='Population of Different Cities', text='Population (Lakhs)')
fig.show()
```
