#### 6.2 K-Means with PCA (3D Plot)

```python
# ==========================================
# IMPORTS
# ==========================================
import numpy as np
import pandas as pd
import plotly.express as px
import requests
from bs4 import BeautifulSoup
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from bokeh.plotting import figure, show, output_notebook
from bokeh.models import ColumnDataSource, HoverTool

# ==========================================
# 1. NumPy Array Operations
# ==========================================
print("--- 1. NumPy Array Operations ---")
# Create 2D array (3,4) with random ints 0-10
arr = np.random.randint(0, 11, size=(3, 4))
print("Original:\n", arr)

# Mean and Standard Deviation
print("Mean:", arr.mean(), "| Std Dev:", arr.std())

# Slice: first 2 rows, last 2 columns
print("Sliced:\n", arr[:2, -2:])

# Reshape to (4, 3)
reshaped = arr.reshape(4, 3)
print("Reshaped:\n", reshaped)

# Element-wise multiplication with same shape
arr2 = np.random.randint(1, 5, size=(4, 3))
print("Multiplied:\n", reshaped * arr2)


# ==========================================
# 2. Feature Engineering & EDA
# ==========================================
print("\n--- 2. Feature Engineering & EDA ---")
np.random.seed(42)
df = pd.DataFrame({
    'Feature1': np.random.uniform(1, 100, 150),
    'Feature2': np.random.uniform(1, 100, 150),
    'Target':   np.random.choice(['A', 'B'], 150)
})

df['Log_Feature1'] = np.log(df['Feature1'])
df['Feature2_Category'] = pd.cut(df['Feature2'], bins=3, labels=['Low','Medium','High'])

print("Correlation Matrix:\n", df[['Feature1','Feature2','Log_Feature1']].corr())

px.scatter(df, x='Feature1', y='Feature2', color='Target',
           title='Feature1 vs Feature2').show()


# ==========================================
# 3. Interactive Bar Chart with Plotly
# ==========================================
print("\n--- 3. Interactive Bar Chart with Plotly ---")
data_pop = {'City': ['Mumbai', 'Delhi', 'Bengaluru', 'Hyderabad', 'Chennai', 'Kolkata'],
            'Population (Lakhs)': [204, 167, 136, 108, 115, 149]}
df_pop = pd.DataFrame(data_pop)

fig = px.bar(df_pop, x='City', y='Population (Lakhs)', color='City',
             title='Population of Different Cities', text='Population (Lakhs)')
fig.show()


# ==========================================
# 4. Interactive Visualizations with Bokeh
# ==========================================
print("\n--- 4. Interactive Visualizations with Bokeh ---")
output_notebook()

# 4.1 Interactive Bar Chart (Sales Figures)
data_sales = {'Product': ['Laptop','Smartphone','Headphones','Smart Watch','Tablet',
                          'Printer','Keyboard','Monitor','External Hard Drive','Webcam'],
              'Sales':   [20000, 24000, 3750, 6000, 7000, 2250, 1800, 4500, 3500, 2000]}
df_sales = pd.DataFrame(data_sales)
source_sales = ColumnDataSource(df_sales)

p1 = figure(x_range=df_sales['Product'], title='Sales Figures of Different Products',
            x_axis_label='Products', y_axis_label='Total Sales ($)',
            height=450, width=850)
p1.vbar(x='Product', top='Sales', width=0.6, source=source_sales, color='steelblue')
p1.add_tools(HoverTool(tooltips=[("Product", "@Product"), ("Sales ($)", "@Sales")]))
show(p1)


# 4.2 Interactive Line Plot (Daily Temperatures)
data_temp = {'Day': ['Monday','Tuesday','Wednesday','Thursday','Friday','Saturday','Sunday'],
             'Temperature': [28, 30, 29, 31, 33, 32, 30]}
df_temp = pd.DataFrame(data_temp)
source_temp = ColumnDataSource(df_temp)

p2 = figure(x_range=df_temp['Day'], title='Daily Temperatures Over a Week',
            x_axis_label='Day', y_axis_label='Temperature (°C)',
            height=450, width=750)
p2.line(x='Day', y='Temperature', source=source_temp, line_width=2, color='blue')
p2.scatter(x='Day', y='Temperature', source=source_temp, size=8, color='red') # FIXED: p.circle deprecated
p2.add_tools(HoverTool(tooltips=[("Day", "@Day"), ("Temperature (°C)", "@Temperature")]))
show(p2)


# 4.3 Interactive Scatter Plot (Iris Dataset)
iris = load_iris()
df_iris = pd.DataFrame(iris.data, columns=iris.feature_names)
source_iris = ColumnDataSource(df_iris)

p3 = figure(title='Relationship: Petal Length vs Petal Width (Iris)',
            x_axis_label='Petal Length (cm)', y_axis_label='Petal Width (cm)',
            height=450, width=700)
p3.scatter(x='petal length (cm)', y='petal width (cm)', 
           source=source_iris, size=8, color='blue', alpha=0.6)
p3.add_tools(HoverTool(tooltips=[("Petal Length", "@{petal length (cm)}"),
                                 ("Petal Width", "@{petal width (cm)}")]))
show(p3)


# ==========================================
# 5. Web Scraping and Data Collection
# ==========================================
print("\n--- 5. Web Scraping and Data Collection ---")
url = "https://quotes.toscrape.com"
response = requests.get(url)
soup = BeautifulSoup(response.text, 'html.parser')

data_quotes = []
for q in soup.find_all('div', class_='quote'):
    data_quotes.append({
        'Quote': q.find('span', class_='text').text,
        'Author': q.find('small', class_='author').text,
        'Tags': ", ".join([t.text for t in q.find_all('a', class_='tag')])
    })

df_quotes = pd.DataFrame(data_quotes)
df_quotes.to_csv('quotes.csv', index=False)
print(df_quotes.head())


# ==========================================
# 6. K-Means Clustering on Iris Dataset
# ==========================================
print("\n--- 6. K-Means Clustering on Iris Dataset ---")

# 6.1 K-Means with Raw Features (3D Plot)
X_scaled = StandardScaler().fit_transform(df_iris)
df_iris['cluster'] = KMeans(n_clusters=3, random_state=42, n_init=10).fit_predict(X_scaled).astype(str)

px.scatter_3d(df_iris, x='petal length (cm)', y='petal width (cm)',
              z='sepal length (cm)', color='cluster',
              title='K-Means on Iris (3D Raw Features)').show()


# 6.2 K-Means with PCA (3D Plot)
pca = PCA(n_components=3)
df_iris[['PC1','PC2','PC3']] = pca.fit_transform(X_scaled)

px.scatter_3d(df_iris, x='PC1', y='PC2', z='PC3', color='cluster', 
              title='K-Means on Iris (PCA)').show()
```
