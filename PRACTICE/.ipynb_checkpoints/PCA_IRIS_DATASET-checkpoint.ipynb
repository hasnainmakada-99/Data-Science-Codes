```python
import pandas as pd, plotly.express as px
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

iris = load_iris()

df = pd.DataFrame(iris.data, columns=iris.feature_names)

X = StandardScaler().fit_transform(df)

df['cluster'] = KMeans(n_clusters=3, random_state=42, n_init=10).fit_predict(X).astype(str)

df[['PC1','PC2','PC3']] = PCA(n_components=3).fit_transform(X)

px.scatter_3d(df, x='PC1', y='PC2', z='PC3', color='cluster', title='K-Means on Iris').show()
```
