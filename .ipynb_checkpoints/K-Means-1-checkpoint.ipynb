```python
# Step 1: Import Necessary Libraries
import numpy as np
import pandas as pd
from sklearn.cluster import KMeans
import plotly.express as px
```

```python
# Step 2: Create DataSet for K Means
# Step 3: Prepare Data for K-Means
data = np.array([2, 4, 10, 12, 3, 20, 11, 25])
df = pd.DataFrame({'Value': data})
print(df)
```

```python
# Step 4: Apply K-Means with Given Initials
X = data.reshape(-1, 1)
print(X)
```

```python
# Step 5: Assign Cluster Labels
initial_centroids = np.array([[3], [4]])

kmeans = KMeans(n_clusters=2, init=initial_centroids, n_init=1, random_state=42)

kmeans.fit(X)
```

```python
# Step 6: Display Final Centroids
df['Cluster'] = kmeans.labels_
print(df)
```

```python
# Step 7: Create Data for 3D Visualization
df['X'] = df['Value']

df['Y'] = np.random.randint(1, 10, len(df))

df['Z'] = np.random.randint(1, 10, len(df))
```

```python
fig = px.scatter_3d(data_frame=df, x='X', y='Y', z='Z', color=df['Cluster'].astype(str), hover_data=['Value', 'Cluster'], title='K Means Clustering Visualization', size_max=600)

fig.show()
```
