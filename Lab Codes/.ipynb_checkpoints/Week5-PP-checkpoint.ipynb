```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import json
```

```python
url = "https://quotes.toscrape.com/"

response = requests.get(url)

print("Status Code:", response.status_code)
```

```python
soup = BeautifulSoup(response.text, "html.parser")

print("Website Title:", soup.title.text)
```

```python
quotes = []

for item in soup.find_all("div", class_="quote"):

    quote = item.find("span", class_="text").text
    author = item.find("small", class_="author").text

    tags = [
        tag.text
        for tag in item.find_all("a", class_="tag")
    ]

    quotes.append({
        "Quote": quote,
        "Author": author,
        "Tags": ", ".join(tags)
    })

print("Number of quotes collected:", len(quotes))
```

```python
df = pd.DataFrame(quotes)

df
```

```python
df.to_csv("quotes.csv", index=False)

print("Data saved successfully in CSV format.")
```

```python
with open("quotes.json", "w", encoding="utf-8") as file:
    json.dump(quotes, file, indent=4, ensure_ascii=False)

print("Data saved successfully in JSON format.")
```

```python
all_quotes = []

page = 1

while True:

    url = f"https://quotes.toscrape.com/page/{page}/"

    response = requests.get(url)
    soup = BeautifulSoup(response.text, "html.parser")

    items = soup.find_all("div", class_="quote")

    if not items:
        break

    for item in items:

        quote = item.find("span", class_="text").text
        author = item.find("small", class_="author").text

        tags = [
            tag.text
            for tag in item.find_all("a", class_="tag")
        ]

        all_quotes.append({
            "Quote": quote,
            "Author": author,
            "Tags": ", ".join(tags)
        })

    page += 1

print("Total quotes collected:", len(all_quotes))
```

```python
df = pd.DataFrame(all_quotes)

df
```

```python
df.to_csv("quotes_dataset.csv", index=False)

with open("quotes_dataset.json", "w", encoding="utf-8") as file:
    json.dump(all_quotes, file, indent=4, ensure_ascii=False)

print("CSV and JSON files created successfully.")
```
